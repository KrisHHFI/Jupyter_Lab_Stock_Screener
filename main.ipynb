```python
import yfinance as yf
from utils.get_stock_info import get_stock_info
from utils.get_today_price_movement import get_today_price_movement
from utils.plot_today_price_movement import plot_today_price_movement
from utils.print_stock_info import print_stock_info
from utils.print_data_collected import print_data_collected
from utils.yfinance_call_tracker import (
    get_yfinance_call_count,
    increment_yfinance_call_count,
    reset_yfinance_call_count,
 )

# Define stock symbol
chosenStock = "aapl"

try:
    reset_yfinance_call_count()
    increment_yfinance_call_count()
    ticker = yf.Ticker(chosenStock)

    intraday_data = get_today_price_movement(chosenStock, ticker=ticker, return_full_data=True)
    intraday_close = intraday_data["Close"]
    stock_info = get_stock_info(chosenStock, ticker=ticker, intraday_data=intraday_data)
    print_stock_info(stock_info)
    plot_today_price_movement(chosenStock, intraday_close)
    print_data_collected(yfinance_call_count=get_yfinance_call_count())
except Exception as e:
    print(f"Error: {e}")
```
